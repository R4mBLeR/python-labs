```python
import pandas as pd
import numpy as np

df = pd.read_csv('students.csv', sep=';')

columns = ['Программирование', 'Мат. анализ', 'Иностранный язык', 'Механика', 'Агила']

grades = df[columns].to_numpy()

avg_per_student = np.mean(grades, axis=1)
avg_per_disc = np.mean(grades, axis=0)

df['Средний балл'] = avg_per_student

best_idx = np.argmax(avg_per_student)
worst_idx = np.argmin(avg_per_student)

print(df[['ФИ', 'Средний балл']])
print()
print(f'Лучший студент: {df.iloc[best_idx]["ФИ"]}')
print(f'Худший студент: {df.iloc[worst_idx]["ФИ"]}')

for subject, avg in zip(columns, avg_per_disc):
    print(f"{subject}: {avg:.2f}")

easiest_disc_idx = np.argmax(avg_per_disc)
hardest_disc_idx = np.argmin(avg_per_disc)

easiest_disc = columns[easiest_disc_idx]
hardest_disc = columns[hardest_disc_idx]

print(f'Легчайшая дисциплина: {easiest_disc} (средний балл: {avg_per_disc[easiest_disc_idx]:.2f})')
print(f'Сложнейшая дисциплина: {hardest_disc} (средний балл: {avg_per_disc[hardest_disc_idx]:.2f})')
```
